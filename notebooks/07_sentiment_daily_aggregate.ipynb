#감성분석 후 데이터가공_first

```python
df_first_Hyundai = pd.read_csv(f'{first_path}/Hyundai_sentiment_result.csv').drop('content', axis=1)
df_first_Kbfg = pd.read_csv(f'{first_path}/Kbfg_sentiment_result.csv').drop('content', axis=1)
df_first_Kia = pd.read_csv(f'{first_path}/Kia_sentiment_result.csv').drop('content', axis=1)
df_first_Lgchem = pd.read_csv(f'{first_path}/Lgchem_sentiment_result.csv').drop('content', axis=1)
df_first_Lgensol = pd.read_csv(f'{first_path}/Lgensol_sentiment_result.csv').drop('content', axis=1)
df_first_Naver = pd.read_csv(f'{first_path}/Naver_sentiment_result.csv').drop('content', axis=1)
df_first_PoscoHoldings = pd.read_csv(f'{first_path}/PoscoHoldings_sentiment_result.csv').drop('content', axis=1)
df_first_SamsungBiologics = pd.read_csv(f'{first_path}/SamsungBiologics_sentiment_result.csv').drop('content', axis=1)
df_first_Samsung = pd.read_csv(f'{first_path}/Samsung_sentiment_result.csv').drop('content', axis=1)
df_first_Skhynix = pd.read_csv(f'{first_path}/Skhynix_sentiment_result.csv').drop('content', axis=1)
```

```python
avg_emotion_Hyundai = df_first_Hyundai.groupby('일자').agg({
    '부정 확률 (0)': 'mean',
    '긍정 확률 (1)': 'mean'
}).reset_index()
avg_emotion_Hyundai['감성'] = avg_emotion_Hyundai.apply(lambda row: '긍정' if row['긍정 확률 (1)'] >= row['부정 확률 (0)'] else '부정', axis=1)
df_first_Hyundai = avg_emotion_Hyundai

avg_emotion_Kbfg = df_first_Kbfg.groupby('일자').agg({
    '부정 확률 (0)': 'mean',
    '긍정 확률 (1)': 'mean'
}).reset_index()
avg_emotion_Kbfg['감성'] = avg_emotion_Kbfg.apply(lambda row: '긍정' if row['긍정 확률 (1)'] >= row['부정 확률 (0)'] else '부정', axis=1)
df_first_Kbfg = avg_emotion_Kbfg

avg_emotion_Kia = df_first_Kia.groupby('일자').agg({
    '부정 확률 (0)': 'mean',
    '긍정 확률 (1)': 'mean'
}).reset_index()
avg_emotion_Kia['감성'] = avg_emotion_Kia.apply(lambda row: '긍정' if row['긍정 확률 (1)'] >= row['부정 확률 (0)'] else '부정', axis=1)
df_first_Kia = avg_emotion_Kia

avg_emotion_Lgchem = df_first_Lgchem.groupby('일자').agg({
    '부정 확률 (0)': 'mean',
    '긍정 확률 (1)': 'mean'
}).reset_index()
avg_emotion_Lgchem['감성'] = avg_emotion_Lgchem.apply(lambda row: '긍정' if row['긍정 확률 (1)'] >= row['부정 확률 (0)'] else '부정', axis=1)
df_first_Lgchem = avg_emotion_Lgchem

avg_emotion_Lgensol = df_first_Lgensol.groupby('일자').agg({
    '부정 확률 (0)': 'mean',
    '긍정 확률 (1)': 'mean'
}).reset_index()
avg_emotion_Lgensol['감성'] = avg_emotion_Lgensol.apply(lambda row: '긍정' if row['긍정 확률 (1)'] >= row['부정 확률 (0)'] else '부정', axis=1)
df_first_Lgensol = avg_emotion_Lgensol

avg_emotion_Naver = df_first_Naver.groupby('일자').agg({
    '부정 확률 (0)': 'mean',
    '긍정 확률 (1)': 'mean'
}).reset_index()
avg_emotion_Naver['감성'] = avg_emotion_Naver.apply(lambda row: '긍정' if row['긍정 확률 (1)'] >= row['부정 확률 (0)'] else '부정', axis=1)
df_first_Naver = avg_emotion_Naver

avg_emotion_PoscoHoldings = df_first_PoscoHoldings.groupby('일자').agg({
    '부정 확률 (0)': 'mean',
    '긍정 확률 (1)': 'mean'
}).reset_index()
avg_emotion_PoscoHoldings['감성'] = avg_emotion_PoscoHoldings.apply(lambda row: '긍정' if row['긍정 확률 (1)'] >= row['부정 확률 (0)'] else '부정', axis=1)
df_first_PoscoHoldings = avg_emotion_PoscoHoldings

avg_emotion_SamsungBiologics = df_first_SamsungBiologics.groupby('일자').agg({
    '부정 확률 (0)': 'mean',
    '긍정 확률 (1)': 'mean'
}).reset_index()
avg_emotion_SamsungBiologics['감성'] = avg_emotion_SamsungBiologics.apply(lambda row: '긍정' if row['긍정 확률 (1)'] >= row['부정 확률 (0)'] else '부정', axis=1)
df_first_SamsungBiologics = avg_emotion_SamsungBiologics

avg_emotion_Samsung = df_first_Samsung.groupby('일자').agg({
    '부정 확률 (0)': 'mean',
    '긍정 확률 (1)': 'mean'
}).reset_index()
avg_emotion_Samsung['감성'] = avg_emotion_Samsung.apply(lambda row: '긍정' if row['긍정 확률 (1)'] >= row['부정 확률 (0)'] else '부정', axis=1)
df_first_Samsung = avg_emotion_Samsung

avg_emotion_Skhynix = df_first_Skhynix.groupby('일자').agg({
    '부정 확률 (0)': 'mean',
    '긍정 확률 (1)': 'mean'
}).reset_index()
avg_emotion_Skhynix['감성'] = avg_emotion_Skhynix.apply(lambda row: '긍정' if row['긍정 확률 (1)'] >= row['부정 확률 (0)'] else '부정', axis=1)
df_first_Skhynix = avg_emotion_Skhynix
```
