```python
df_first_Hyundai = pd.read_csv(base_first_1)[columns_to_keep]
df_first_Hyundaidf_first_Kbfg = pd.read_csv(base_first_2)[columns_to_keep]
df_first_Kia = pd.read_csv(base_first_3)[columns_to_keep]
df_first_Lgchem = pd.read_csv(base_first_4)[columns_to_keep]
df_first_Lgensol = pd.read_csv(base_first_5)[columns_to_keep]
df_first_Naver = pd.read_csv(base_first_6)[columns_to_keep]
df_first_PoscoHoldings = pd.read_csv(base_first_7)[columns_to_keep]
df_first_SamsungBiologics = pd.read_csv(base_first_8)[columns_to_keep]
df_first_Samsung = pd.read_csv(base_first_9)[columns_to_keep]
df_first_Skhynix = pd.read_csv(base_first_10)[columns_to_keep]
```

```python
df_first_Hyundai.head()
```

**Combining titles, keywords, and content**

```python
all_first_dfs = [
    df_first_Hyundai,
    df_first_Kbfg,
    df_first_Kia,
    df_first_Lgchem,
    df_first_Lgensol,
    df_first_Naver,
    df_first_PoscoHoldings,
    df_first_SamsungBiologics,
    df_first_Samsung,
    df_first_Skhynix
]

for i, df in enumerate(all_first_dfs):
    if not df.empty:
        df['content'] = df['제목'].astype(str) + ' ' + \
                        df['키워드'].astype(str) + ' ' + \
                        df['본문'].astype(str)
        df.drop(columns=['제목','키워드','본문'], errors='ignore', inplace=True)
```
