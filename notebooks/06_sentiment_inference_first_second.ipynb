#Predict first file

```python
def predict_sentiment(texts, model, tokenizer, device, batch_size=16):
    predictions = []
    probabilities = []

    for i in range(0, len(texts), batch_size):
        batch_texts = texts[i: i + batch_size]

        inputs = tokenizer(batch_texts, truncation=True, padding=True,
                           max_length=512, return_tensors='pt').to(device)

        with torch.no_grad():
            outputs = model(**inputs)

        logits = outputs.logits
        probs = softmax(logits, dim=1).cpu().numpy()
        preds = torch.argmax(logits, dim=1).cpu().numpy()

        predictions.extend(preds)
        probabilities.extend(probs)

    return predictions, probabilities
```

```python
csv_files = {
    'Hyundai': 'df_first_Hyundai.csv',
    'Kbfg': 'df_first_Kbfg.csv',
    'Kia': 'df_first_Kia.csv',
    'Lgchem': 'df_first_Lgchem.csv',
    'Lgensol': 'df_first_Lgensol.csv',
    'Naver': 'df_first_Naver.csv',
    'PoscoHoldings': 'df_first_PoscoHoldings.csv',
    'Samsung': 'df_first_Samsung.csv',
    'SamsungBiologics': 'df_first_SamsungBiologics.csv',
    'Skhynix': 'df_first_Skhynix.csv'
}

base_path = '/content/drive/MyDrive/dataset/df_first/'
save_path = '/content/drive/MyDrive/result/first_file'

os.makedirs(save_path, exist_ok=True)
```

```python
for company, filename in tqdm(csv_files.items(), desc='기업별 뉴스 감성분석'):
    print(f"▶ {company} 처리 중...")

    try:
        df = pd.read_csv(os.path.join(base_path, filename))

        news_texts = df['content'].fillna("").tolist()
        dates = df['일자'].tolist()

        preds, probs = predict_sentiment(news_texts, model, tokenizer, device)

        result_df = pd.DataFrame({
            '일자': dates,
            'content': news_texts,
            '예측 라벨': preds,
            '부정 확률 (0)': [f"{p[0]:.3f}" for p in probs],
            '긍정 확률 (1)': [f"{p[1]:.3f}" for p in probs],
            '감성': ['긍정' if label == 1 else '부정' for label in preds]
        })

        output_file = os.path.join(save_path, f'{company}_sentiment_result.csv')
        result_df.to_csv(output_file, index=False, encoding='utf-8-sig')
        print(f"{company} 완료 → 저장: {output_file}")

    except Exception as e:
        print(f"{company} 처리 중 오류 발생: {e}")
```

```python
def copy_first_file(base_path,destination_path):
  folder_name=os.path.basename(base_path)
  destination_path=os.path.join(destination_path,folder_name)

  shutil.copytree(base_path,destination_path)
```

```python
base_path='/content/drive/MyDrive/result/first_file'
destination_path='/content/drive/MyDrive/share_drive/박박박사/emotion_analysis'

copy_first_file(base_path,destination_path)
```
