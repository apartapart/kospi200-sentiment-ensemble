#Combine

```python
import torch
from sklearn.model_selection import train_test_split
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments, AutoModel
from torch.utils.data import Dataset
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
```

**Label ebcidubg**

```python
text_column_name = 'content'

df_pos['label'] = 1

df_neg['label'] = 0
```

**Mix data**

```python
combined_df = pd.concat([df_pos, df_neg]).sample(frac=1, random_state=42).reset_index(drop=True)
```

**get(0,0)} : used to express '0' even values without**

```python
all_texts = combined_df[text_column_name].tolist()
all_labels = combined_df['label'].tolist()

print(f"Number of Total data: {len(all_texts)}")
print(f"Pos Lable ({combined_df['label'].value_counts().get(1, 0)}), Neg Lable ({combined_df['label'].value_counts().get(0, 0)})")
```

**Verification : Data for interim checks used to prevent overfitting**

```python
train_texts, temp_texts, train_labels, temp_labels = train_test_split(
    all_texts, all_labels, test_size=0.2, random_state=42, stratify=all_labels
)

val_texts, test_texts, val_labels, test_labels = train_test_split(
    temp_texts, temp_labels, test_size=0.5, random_state=42, stratify=temp_labels
)
```

```python
print(f"Number of Train data : {len(train_texts)}")
print(f"Number of Validation data: {len(val_texts)}")
print(f"Number of Test data: {len(test_texts)}")
```

#Load tokenizer and model

https://klue-benchmark.com/

https://medium.com/@kanghee.lee/klue-korean-language-understanding-evaluation-d06850888d41

**A standard benchmark dataset and framework for systematically evaluating and comparing the performance of various Korean NLP models, including KoBERT**

```python
MODEL_NAME = "klue/bert-base"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
```

**num_labels : Positives and negatives '2'**

```python
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=2)
```

#Define custom dataset classes

```python
class StockSentimentDataset(Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {key: torch.tensor(val[idx]) for key, val in self.encodings.items()}
        item['labels'] = torch.tensor(self.labels[idx])
        return item

    def __len__(self):
        return len(self.labels)
```

#Text Preprocessing

```python
train_encodings = tokenizer(train_texts, truncation=True, padding=True, max_length=512)
val_encodings = tokenizer(val_texts, truncation=True, padding=True, max_length=512)
test_encodings = tokenizer(test_texts, truncation=True, padding=True, max_length=512)

train_dataset = StockSentimentDataset(train_encodings, train_labels)
val_dataset = StockSentimentDataset(val_encodings, val_labels)
test_dataset = StockSentimentDataset(test_encodings, test_labels)
```

# Calculate Evaluation Indicators for Binary Classification

```python
def compute_metrics(pred):
    labels = pred.label_ids
    preds = pred.predictions.argmax(-1)

    precision, recall, f1, _ = precision_recall_fscore_support(labels, preds, average='binary')
    acc = accuracy_score(labels, preds)
    return {
        'accuracy': acc,
        'f1': f1,
        'precision': precision,
        'recall': recall
    }
```

```python
output_dir = '/content/drive/MyDrive/analysis'
```

#Training Configuration

```python
training_args = TrainingArguments(
    output_dir=output_dir,
    num_train_epochs=5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    warmup_steps=50,
    weight_decay=0.01,
    learning_rate=5e-5,
    logging_dir='./logs',
    logging_strategy='steps',
    logging_steps=50,
    eval_strategy='epoch',
    save_strategy='epoch',
    load_best_model_at_end=True,
    metric_for_best_model='f1',
    greater_is_better=True,
    dataloader_num_workers=4,
    seed=42
)
```

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,    #Prevent overfit
    compute_metrics=compute_metrics,
)
```
