#Change object data -> int data

**1. datatime**

```python
df['date'] = pd.to_datetime(df['date'])
print(f"orginal date' column type: {df['date'].dtype}")

df['y'] = df['date'].dt.year
df['m'] = df['date'].dt.month
df['d'] = df['date'].dt.day

print(f"'y' column type: {df['y'].dtype}")
print(f"'m' column type: {df['m'].dtype}")
print(f"'d' column type: {df['d'].dtype}")

existing_columns = df.columns.tolist()
for col in ['y', 'm', 'd']:
    if col in existing_columns:
        existing_columns.remove(col)

new_column_order = ['y', 'm', 'd'] + existing_columns
df = df[new_column_order]
```

```python
df.drop(columns=['date'], inplace=True)
```

```python
df.head()
```

**2. volumn(M)**

```python
df['volume_M'] = df['volume_M'].str.replace('M', '', regex=False)
df['volume_M'] = df['volume_M'].str.replace('K', '', regex=False)
df['volume_M'] = df['volume_M'].astype(float)
```

**3. change %**

```python
df['change'] = df['change'].str.replace('%', '', regex=False)
df['change'] = df['change'].astype(float)
```
